# Регрессия (Regression)

## Функция регрессии

$$
y = f(x_1, x_2, ..., x_n) + \sigma
$$

- $f$ - функция зависимости

- $x_i$ - признаки (features)

- $y$ - целевая переменная (target)

- $\sigma$ - ошибка (шум)


## Математическая формула

$$
y = a + bx + \sigma
$$

- $y$ - целевая переменная (target)

- $x$ - независимоя переменная (фактор)

- $a$ - свободный член (intercept)

- $b$ - коэффициент наклона (slope)

- $\sigma$ - случайная ошибка (разница между реальным и предсказанным значением)


$$
\sigma = \hat{y} - y
$$



$$
y = a + bx + \sigma
$$


$$
b = \frac{n\sum{xy} - \sum{x}\sum{y}}{n\sum{x^2} - (\sum{x})^2}
$$


$$
a = \frac{\sum{y} - b\sum{x}}{n}
$$

```python
X = np.array([1, 2, 3, 4, 5])
y = np.array([53, 59, 65, 72, 76])
```

```python
n = len(X)

b = (n * np.sum(X*y) - np.sum(X)*np.sum(y))/(n*np.sum(X**2) - np.sum(X) ** 2)
a = (np.sum(y) - b * np.sum(X))/n
```

```python
print(f'Уравнение у = {a:.2f} + {b:.2f} * x')
```

```python
y_pred = a + b*X
y_pred
```

```python
plt.scatter(X, y, color='red', label='Datas', zorder=2)
plt.plot(X, y, color='blue', label='Linear Regression', zorder=1)
plt.xlabel('Hours')
plt.ylabel('Marks')
plt.legend()

plt.show()
```

```python
while True:
    try:
        x_new = float(input('Enter hours'))
        if x_new == 0:

            print('Exit')
            break
        y_new = a + b*x_new
        print(f"Prediction: y = {y_new:.2f}")
        plt.scatter(X, y, color='red', label='Datas', zorder=2)
        plt.scatter(x_new, y_new, color='silver', label='Our prediction', zorder=2, s=100)
        plt.plot(X, y, color='blue', label='Linear Regression', zorder=1)
        plt.title(f'Prediction for {x_new:.2f} hours')
        plt.xlabel('Hours')
        plt.ylabel('Marks')
        plt.legend()
        plt.show()
    except ValueError:
        print('Enter the number')
        
```

```python
from sklearn.linear_model import LinearRegression
```

```python
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([53, 59, 65, 72, 76])
```

```python
print(X)
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
a = model.intercept_
b = model.coef_[0]
```

```python
print(f'Уравнение у = {a:.2f} + {b:.2f} * x')
```

```python
y_pred = model.predict(X)
```
