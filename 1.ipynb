```python
import matplotlib 
import matplotlib.pyplot as plt
import numpy as np

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

temperature = [18, 20, 22, 21, 24, 26, 23]

plt.figure(figsize=(10, 5))
plt.title('Weekly Temperature')
plt.xlabel('Days')
plt.ylabel('Temperature')
plt.grid()

plt.plot(days,temperature, linestyle='--', marker='o')

y_range = max(temperature) - min(temperature)
offset = y_range * 0.02


for x, y in zip(days, temperature):

    plt.text(x, y + offset, f'{y}', ha='center',va='bottom')




plt.show()




```

```python
import random
import numpy as np
import matplotlib.pyplot as plt

languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]

max_idx = np.argmax(students)
max_x = students[max_idx]
max_y = languages[max_idx]



plt.figure(figsize=(10, 5))
plt.barh(languages, students, color=['red', 'blue', 'green', 'orange', 'yellow'])


plt.scatter(
    max_x - 0.5,
    max_y,
    color='grey',
    s=100,
    label=f'Max X ({max_x}, {max_y})',
)




plt.title('Student by Programming Language')
plt.xlabel('Number of Students')
plt.ylabel('Languages')

for i, v in enumerate(students):
  plt.text(v + 0.2, i, str(v), va='center') 

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]



sbar = plt.scatter(study_hours, scores, c=ages)

cbar = plt.colorbar(sbar)
cbar.set_label('Age')


max_y = max(scores)
max_x = study_hours[scores.index(max_y)]

plt.annotate(
    "Highest score",          
    xy=(max_x, max_y),       
    xytext=(0, 10),          
    textcoords="offset points",  
    ha="center",               
    fontweight="bold"         
)



plt.title('Study hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid()
plt.show()
```
