```python
import matplotlib 
import matplotlib.pyplot as plt
import numpy as np

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

temperature = [18, 20, 22, 21, 24, 26, 23]

plt.figure(figsize=(10, 5))
plt.title('Weekly Temperature')
plt.xlabel('Days')
plt.ylabel('Temperature')
plt.grid()

plt.plot(days,temperature, linestyle='--', marker='o')

y_range = max(temperature) - min(temperature)
offset = y_range * 0.02


for x, y in zip(days, temperature):

    plt.text(x, y + offset, f'{y}', ha='center',va='bottom')




plt.show()




```

```python
import random
import numpy as np
import matplotlib.pyplot as plt

languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]

max_idx = np.argmax(students)
max_x = students[max_idx]
max_y = languages[max_idx]



plt.figure(figsize=(10, 5))
plt.barh(languages, students, color=['red', 'blue', 'green', 'orange', 'yellow'])


plt.scatter(
    max_x - 0.5,
    max_y,
    color='grey',
    s=100,
    label=f'Max X ({max_x}, {max_y})',
)




plt.title('Student by Programming Language')
plt.xlabel('Number of Students')
plt.ylabel('Languages')

for i, v in enumerate(students):
  plt.text(v + 0.2, i, str(v), va='center') 

plt.legend()
plt.show()
```
