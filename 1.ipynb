```python
import matplotlib 
import matplotlib.pyplot as plt
import numpy as np

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

temperature = [18, 20, 22, 21, 24, 26, 23]

plt.figure(figsize=(10, 5))
plt.title('Weekly Temperature')
plt.xlabel('Days')
plt.ylabel('Temperature')
plt.grid()

plt.plot(days,temperature, linestyle='--', marker='o')

y_range = max(temperature) - min(temperature)
offset = y_range * 0.02


for x, y in zip(days, temperature):

    plt.text(x, y + offset, f'{y}', ha='center',va='bottom')




plt.show()




```
